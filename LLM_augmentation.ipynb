```python
import pandas as pd

df_bruto = pd.read_csv("data/raw_datasets/mbti_1.csv")

# (Opcional) Uma limpezinha bem leve apenas de URLs antes da IA, se quiser ajudar o Llama
preprocessor_leve = Preprocessor(texts_column='posts')
df_bruto = preprocessor_leve.remove_urls(df_bruto)

# 2. Instancia e roda o LlamaAugmenter PRIMEIRO (no texto com gramática)
augmenter = LlamaAugmenter(texts_column='posts', label_column='type')
df_aumentado_bruto = augmenter.augment_minority_classes(df_bruto)

# 3. AGORA SIM: Limpeza Pesada em TUDO (textos originais + gerados pela IA)
print("Iniciando a limpeza pesada (NLP)...")
preprocessor_pesado = Preprocessor(texts_column='posts')
df_final_limpo = preprocessor_pesado.preprocess_complete(df_aumentado_bruto)

# Salva o resultado final pronto para o Word2Vec
df_final_limpo.head()
```
