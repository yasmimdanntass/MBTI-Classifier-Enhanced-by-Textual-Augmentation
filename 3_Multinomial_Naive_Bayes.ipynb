```python
import pandas as pd

from pipeline.multinomial_naive_bayes import MultinomialNaiveBayesClassifier
```

```python
dataframe = pd.read_csv("data/raw_datasets/mbti_1.csv")
dataframe.head()
```

```python
# Cria as colunas dos eixos

dataframe["is_introvert"] = dataframe["type"].str[0] == "I"
dataframe["is_intuitive"] = dataframe["type"].str[1] == "N"
dataframe["is_feeling"] = dataframe["type"].str[2] == "F"
dataframe["is_judging"] = dataframe["type"].str[3] == "J"
```

```python
dataframe[
    [
        "type",
        "is_introvert",
        "is_intuitive",
        "is_feeling",
        "is_judging"
    ]
].head()
```

```python
# Cria os 4 classificadores

classifier_ie = MultinomialNaiveBayesClassifier()
classifier_ns = MultinomialNaiveBayesClassifier()
classifier_ft = MultinomialNaiveBayesClassifier()
classifier_jp = MultinomialNaiveBayesClassifier()
```

```python
# Treinar o eixo I/E

x_test_ie, y_test_ie = classifier_ie.train(dataframe, "is_introvert")

metrics_ie = classifier_ie.evaluate(x_test_ie, y_test_ie)
```

```python
# Treina o eixo N/S

x_test_ns, y_test_ns = classifier_ns.train(dataframe, "is_intuitive")

metrics_ns = classifier_ns.evaluate(x_test_ns, y_test_ns)
```

```python
# Treina o eixo F/T

x_test_ft, y_test_ft = classifier_ft.train(dataframe, "is_feeling")

metrics_ft = classifier_ft.evaluate(x_test_ft, y_test_ft)
```

```python
# Treina o eixo J/P

x_test_jp, y_test_jp = classifier_jp.train(dataframe, "is_judging")

metrics_jp = classifier_jp.evaluate(x_test_jp, y_test_jp)
```

```python
def print_metrics(axis_name, metrics):
    accuracy, precision, recall, f1, report = metrics

    print(f"Eixo: {axis_name}")
    print(f"Accuracy: {accuracy:.4f}")
    print(f"Precision macro: {precision:.4f}")
    print(f"Recall macro: {recall:.4f}")
    print(f"F1 macro: {f1:.4f}")
    print("\nClassification report:")
    print(report)
```

```python
print_metrics("I/E", metrics_ie)
print_metrics("N/S", metrics_ns)
print_metrics("F/T", metrics_ft)
print_metrics("J/P", metrics_jp)
```
